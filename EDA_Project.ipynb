```python
# Import Required Libraries

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder

# Display plots inside notebook
%matplotlib inline

# Improve plot appearance
sns.set_style("whitegrid")
```

```python
# Load Dataset
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/Males.csv"
df = pd.read_csv(url)
```

```python
# Display first five rows
df.head()
```

```python
# Dataset Information

print("Shape of Dataset:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)
```

```python
# Basic Statistical Analysis

df.describe(include='all')
```

```python
# Additional Dataset Information

df.info()
```

```python
# Check Missing Values

print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
# Handle Missing Values

for column in df.columns:

    if df[column].dtype == "object":
        df[column].fillna(df[column].mode()[0], inplace=True)
    else:
        df[column].fillna(df[column].mean(), inplace=True)

print("Missing values after handling:")
print(df.isnull().sum())
```

```python
# Check Duplicate Rows

print("Duplicate Rows:", df.duplicated().sum())
```

```python
# Remove Duplicate Rows

df.drop_duplicates(inplace=True)

print("Dataset Shape After Removing Duplicates:")
print(df.shape)
```

```python
# Clean Column Names

df.columns = df.columns.str.strip()

print(df.columns)
```

```python
# Data Transformation using Label Encoding

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for column in df.columns:
    if df[column].dtype == "object":
        df[column] = le.fit_transform(df[column])

print("Data Types After Transformation:")
print(df.dtypes)
```

```python
# Display Cleaned Dataset

df.head()
```

```python
# Univariate Analysis - Histogram

# Display histograms for all numerical columns
df.hist(figsize=(15, 10), bins=15)

plt.suptitle("Histograms of Numerical Features", fontsize=16)
plt.show()
```

```python
# Remove unnecessary ID column

if 'rownames' in df.columns:
    df.drop('rownames', axis=1, inplace=True)

df.head()
```

```python
# Univariate Analysis - Box Plot

plt.figure(figsize=(10, 6))

sns.boxplot(data=df[['nr', 'year', 'exper', 'wage']])

plt.title("Box Plot of Selected Numerical Variables")
plt.xlabel("Variables")
plt.ylabel("Values")

plt.show()
```

```python
# Univariate Analysis - Density Plot

plt.figure(figsize=(8,5))

sns.kdeplot(df['wage'], fill=True)

plt.title("Density Plot of Wage")
plt.xlabel("Wage")

plt.show()
```

```python
# Bivariate Analysis - Experience vs Wage

plt.figure(figsize=(8, 5))

sns.scatterplot(x='exper', y='wage', data=df)

plt.title("Relationship Between Experience and Wage")
plt.xlabel("Experience")
plt.ylabel("Wage")

plt.show()
```

```python
# Bivariate Analysis - Year vs Wage

plt.figure(figsize=(8, 5))

sns.scatterplot(x='year', y='wage', data=df)

plt.title("Relationship Between Year and Wage")
plt.xlabel("Year")
plt.ylabel("Wage")

plt.show()
```

```python
# Bivariate Analysis - Union Status vs Wage

plt.figure(figsize=(8, 5))

sns.boxplot(x='union', y='wage', data=df)

plt.title("Wage Distribution by Union Status")
plt.xlabel("Union Status (0 = No, 1 = Yes)")
plt.ylabel("Wage")

plt.show()
```

```python
# Multivariate Analysis - Correlation Heatmap

plt.figure(figsize=(12, 8))

correlation = df.corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap of Variables")

plt.show()
```

```python
# Multivariate Analysis - Experience, Wage and Union Status

plt.figure(figsize=(9, 6))

sns.scatterplot(
    x='exper',
    y='wage',
    hue='union',
    data=df,
    palette='Set1'
)

plt.title("Experience vs Wage by Union Status")
plt.xlabel("Experience")
plt.ylabel("Wage")
plt.legend(title="Union Status")

plt.show()
```

```python
# Multivariate Analysis - Pair Plot

selected_columns = ['exper', 'wage', 'year', 'union']

sns.pairplot(
    df[selected_columns],
    hue='union',
    diag_kind='hist'
)

plt.show()
```
